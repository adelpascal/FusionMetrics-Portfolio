Kenapa Ini Seru?

```python
import random

def ramalan_mingguan():
    hari = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat", "Sabtu", "Minggu"]
    nasib = ["Dapat kopi gratis", "Meeting membosankan", "Pulang cepat", "Bos lagi baik", "Fokus kerja"]

    print("=== RAMALAN NASIB MINGGU INI ===")
    
    for d in hari:
        pilihan = random.choice(nasib)
        print(d + ": " + pilihan)

# Jalankan mesinnya!
ramalan_mingguan()
```

Apa yang terjadi di sini? (The Logic Unpack)
for d in hari:: Python akan melihat daftar hari. Dia mulai dari "Senin", lalu masuk ke dalam kode di bawahnya.

Si Asisten Beraksi: Di hari Senin, Python memanggil random.choice(nasib). Dapatlah satu pilihan (misal: "Pulang cepat").

Putar Lagi: Setelah Senin selesai, Python tidak berhenti. Dia balik lagi ke atas, ambil "Selasa", lalu kocok lagi pilihannya.

Selesai: Dia akan terus mengulang sampai "Minggu" tercapai.

Hasilnya di Layarmu (Contoh):
=== RAMALAN NASIB MINGGU INI ===
Senin: Meeting membosankan
Selasa: Bos lagi baik
Rabu: Dapat kopi gratis
Kamis: Fokus kerja
Jumat: Pulang cepat
Sabtu: Dapat kopi gratis
Minggu: Bos lagi baik

Kenapa ini "Sakti"?
Bayangkan kalau hari itu isinya bukan 7 hari, tapi 365 hari (satu tahun). Kamu tidak perlu mengetik 365 baris. Kamu cukup pakai 2 baris kode for tadi, dan Python akan menyelesaikan laporannya dalam waktu kurang dari 0.1 detik.

Inilah cara kerja "otomatisasi" yang bisa menyelamatkan waktumu di kantor. Kamu jadi punya lebih banyak waktu untuk menjadi "Statue" yang tenang sementara komputer bekerja keras untukmu.

Step-by-Step Cara Python Menghasilkan Jawaban Itu:

Python membaca daftar hari (ada 7 item) dan daftar nasib (ada 5 item).

2. Masuk ke Putaran (The Loop)
Kalimat for d in hari: itu artinya: "Ambil satu nama dari daftar hari, simpan di kotak 'd', lalu kerjakan tugas di bawahnya."

Putaran 1 (Senin):

Rahasia di Balik random.choice(nasib)
Bayangkan di atas meja ada 5 kartu nasib yang tertutup. Isinya: Kopi, Meeting, Pulang Cepat, Bos Baik, dan Fokus.

Langkah 1: Hari Senin Maju
Python mengambil satu orang dari daftar hari, yaitu Senin. Sekarang tangan Python memegang kartu nama bertuliskan "Senin".

Langkah 2: Proses "Mendapatkan" Nasib (Ini Intinya!)
Baris kode pilihan = random.choice(nasib) itu adalah perintah ke tangan Python untuk:

Mengacak 5 kartu di meja tadi.

Mengambil satu secara buta.

Misal, yang terambil adalah kartu "Meeting".

Sekarang, di tangan kiri Python ada "Senin", di tangan kanan ada "Meeting".

Langkah 3: Menempelkan Nasib
Baris print(d + ": " + pilihan) adalah perintah untuk:

Menempelkan isi tangan kiri dan tangan kanan, lalu menunjukkannya ke layar: "Senin: Meeting".

SETELAH DICETAK, kartu "Meeting" tadi dikembalikan lagi ke meja dan diaduk lagi. Kartu nama "Senin" dibuang karena sudah selesai.

Bagaimana dia mendapatkan nasib kedua (Selasa)?
Inilah keajaiban Loop (for d in hari):

Setelah langkah ke-3 selesai, Python tidak berhenti. Dia melihat daftar hari lagi. "Oh, masih ada Selasa!"

Langkah Ulang: Python memegang kartu nama "Selasa" di tangan kirinya.

Proses Acak Lagi: Dia merogoh lagi ke meja yang berisi 5 kartu nasib tadi (ingat, kartu "Meeting" tadi sudah dikembalikan, jadi kartu nasibnya tetap lengkap 5).

Nasib Baru: Dia ambil satu secara acak lagi. Duar! Kali ini yang terambil kartu "Bos lagi baik".

Cetak: Dia tunjukkan ke layar: "Selasa: Bos lagi baik".

Kenapa logikamu terasa sulit menerima ini?
Mungkin karena kamu berpikir nasib itu sudah ditentukan dari awal untuk semua hari. Padahal, nasib itu baru "diciptakan" (diacak) tepat saat giliran si hari itu tiba.

Saat Senin tiba, Python baru mengocok dadu.

Saat Selasa tiba, Python mengocok dadu lagi.

Saat Rabu tiba, Python mengocok dadu sekali lagi.

Jadi, Python tidak memberikan satu paket nasib sekaligus. Dia memberikan nasib satu per satu sambil berjalan melewati daftar hari.

Apakah ini yang bikin ganjal?
Coba bayangkan ini: Kamu punya 7 teman. Tiap teman datang ke kamu, kamu lempar koin.

Teman ke-1 datang -> Lempar koin -> Hasilnya "Angka".

Teman ke-2 datang -> Lempar koin lagi -> Hasilnya "Gambar".

Python melakukan hal yang sama. Dia "melempar koin" (memanggil random.choice) setiap kali ada "teman" (hari) baru yang datang di baris for.

1. Jawaban Teknis: "The Seed" (Akar Angka)

```python
import math

def hitung_panjang_einstein(panjang_asli, kecepatan):
    c = 300000  # Kecepatan cahaya (km/s)
    # Rumus Lorentz Contraction (ada di halaman 13 paper Einstein)
    panjang_baru = panjang_asli * math.sqrt(1 - (kecepatan**2 / c**2))
    return panjang_baru

# Kalau kamu gerak pelan (kayak naik motor), panjangnya tetap sama
print(hitung_panjang_einstein(100, 0.01)) 

# Kalau kamu gerak 90% kecepatan cahaya (270.000 km/s)
print(hitung_panjang_einstein(100, 270000)) 
# Hasilnya bakal sekitar 43 cm!
```

Kuncinya ada di "Mesin Penghitung" yang namanya math.sqrt(1 - (kecepatan**2 / c**2)).

Lalu dikuadratkan: hasilnya makin mendekati nol.Rumusnya jadi: 100 * akar(1 - 0).100 * akar(1) = 100 * 1 = 100.Kesimpulan: Kalau kecepatanmu kecil, angka pembaginya jadi hampir nol. Makanya hasilnya tetap 100. Fisika Newton masih berlaku di sini.

3. Bagaimana Bisa Jadi 43? (Contoh: 90% Cahaya)Sekarang kita masukkan angka "Dewa" Einstein:panjang_asli = 100kecepatan = 270.000 (90% dari kecepatan cahaya)c = 300.000

Langkah hitungnya pelan-pelan:Bagian Pecahan: kecepatan / c adalah $270.000 / 300.000 = \mathbf{0.9}$.Kuadratkan: $0.9 \times 0.9 = \mathbf{0.81}$.Kurangi dari 1: $1 - 0.81 = \mathbf{0.19}$.Akar Kuadrat (math.sqrt): Akar dari $0.19$ adalah sekitar $\mathbf{0.435}$.Hasil Akhir: $100 \times 0.435 = \mathbf{43.5}$!3.

Di sinilah letak kejeniusan Einstein yang ada di paper halaman 13 itu. Dia membuktikan kalau alam semesta punya "kecepatan maksimum".

Kenapa harus ada angka 1 di rumus 1 - (v^2/c^2)?
Itu adalah cara matematika untuk bilang: "Sisa".

Kalau kamu belum gerak, sisa "ruang"-mu masih 100% (angka 1).

Begitu kamu gerak, "jatah" ruangmu dikurangi oleh kecepatanmu sendiri.

```python
import math

def hitung_massa_einstein(massa_asli, kecepatan):
    c = 300000
    # Rumus Einstein: Massa baru = Massa asli / faktor Lorentz
    faktor = math.sqrt(1 - (kecepatan**2 / c**2))
    massa_baru = massa_asli / faktor
    return massa_baru

# 1. Massa HP saat diam
print("Massa diam:", hitung_massa_einstein(200, 0)) # Hasil: 200

# 2. Massa HP saat 99% kecepatan cahaya (297.000 km/s)
print("Massa saat ngebut:", hitung_massa_einstein(200, 297000)) 
# Hasilnya: sekitar 1417 gram (Naik 7x lipat!)
```

Konsep: "Makin Ngebut, Makin Berat"
Ini adalah bagian yang paling susah diterima logika manusia biasa. Einstein bilang:

"Benda yang bergerak itu massanya (beratnya) bertambah."

Contoh Gampang:
Bayangkan kamu punya HP seberat 200 gram.

Kalau HP itu diam di meja, beratnya ya 200 gram.

Tapi kalau HP itu kamu tembakkan dengan kecepatan 99% cahaya, beratnya di mata orang yang melihat bukan lagi 200 gram, tapi bisa jadi berton-ton!

Kenapa?
Karena kamu butuh Energi yang luar biasa besar untuk mendorong benda sampai secepat itu. Nah, menurut Einstein, energi yang kamu pakai untuk mendorong itu nggak hilang, tapi malah "berubah wujud" jadi Massa tambahan buat benda itu.
